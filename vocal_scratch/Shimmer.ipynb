# Shimmer

## $ shimmer\_rel = \frac{\frac{1}{N-1} \sum{i=1}^{N-1} |A_i - A_{i-1}|}{Mean(A)} $

## $ shimmer\_percent = shimmer\_rel \times 100 $

```python
import numpy as np
```

```python
def shimmer(amplitudes):

    if (len(amplitudes)<3):

        return {}

    # absolute amplitude diff

    diffs = np.abs(np.diff(amplitudes))

    # mean diff, the 1/(N-1)

    shimmer_abs = np.mean(diffs)

    # shimmer_rel

    shimmer_rel = shimmer_abs / np.mean(amplitudes)

    # shimmer percentage

    shimmer_percent = shimmer_rel * 100

    return {float(shimmer_abs),
           float(shimmer_rel), 
           float(shimmer_percent)
           }
    
```

```python
A = np.array([0.50, 0.48, 0.52, 0.47, 0.51])
print(shimmer(A))
```
