# Jitter

```python
import numpy as np
```

## $ jitter\_rel = \frac{\frac{1}{N-1} \sum_{i=0}^{N-1} |T_i - T_{i+1}| }{Mean(T)} $

## $ jitter\_percent = jitterrel \times 100 $

```python
def jitter(F0):

    """
    F0 to T
        1.0/ F0(i)
    """

    Periods = 1.0/F0

    # absolute diff

    diffs = np.abs(np.diff(Periods))

    # Mean (T)

    jitter_abs = np.mean(diffs)

    # jiter rel

    jitter_rel = jitter_abs / np.mean(Periods) 

    # Jitter percentage

    jitter_percent = jitter_rel * 100

    return {
        float(jitter_abs),
        float(jitter_rel),
        float(jitter_percent)
    }
    

    
```

```python
import matplotlib.pyplot as plt
f0 = np.array([150, 152, 149, 151, 150])

plt.figure(figsize=(14, 4))
plt.plot(f0)
plt.show()

print(jitter(f0))
```

```python
import librosa

audio, sr = librosa.load("Jane.mp3", sr = 16000)
f0, voiced_flag, voiced_probs = librosa.pyin(audio, fmin=librosa.note_to_hz('C2'), fmax=librosa.note_to_hz('C6'))

f0_voiced = f0[voiced_flag]
```

```python
jitter(f0_voiced)
```
