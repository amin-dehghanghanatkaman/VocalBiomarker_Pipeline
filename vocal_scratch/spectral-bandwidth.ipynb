## Spectral Bandwidth

```python
import numpy as np
```

```python
def spectralBandwidth(signal, sr=16000):

    X = np.fft.rfft(signal)
    mag = np.abs(X)
    print(f"The Magnitude of short time Fourier is : {mag}")
    print("*"*80)
    freq = np.arange(len(mag))

    #Calculating SC

    sc = np.sum(freq*mag) / np.sum(mag)

    # Now Variance
    variance = np.sum(((freq-sc)**2 * mag) /(np.sum(mag)))
    #Band Width

    bw = np.sqrt(variance)
    print("bnad width: ", bw)

    bin_width = sr / len(signal)
    return bw * bin_width
    

    
    

    
    
```

```python
signal = np.array([1.0, 0.5, -0.5, -1.0, -0.5, 0.5, 1.0, 0.5])
spectralBandwidth(signal, sr=16000)
```

```python
import librosa

sr = 16000

audio_path = "test_healthy_voice.wav"

audio, sr = librosa.load(audio_path, sr=sr , mono=True )
print(audio)
```

```python
spectral_bandwidth = librosa.feature.spectral_bandwidth(y=audio, sr = sr)
print(spectral_bandwidth)

# Mean Spectral Bandwidth
print("#"*80)
print(np.mean(spectral_bandwidth))
```
