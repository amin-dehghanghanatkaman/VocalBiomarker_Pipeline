```python
import numpy as np
```

```python
def zcr(signal):

    N = len(signal)

    sign = np.sign(signal)

    diff = np.diff(sign)

    abs_ = np.abs(diff)
    sum_ = np.sum(abs_)

    zcr = float((1/(2*(N-1)))* sum_)

    return zcr
```

```python
x = [0.3, -0.2, -0.1, 0.4, 0.5]

zcr(x)
```

```python
import librosa
```

```python
sr = 6000

audio_path = "test_healthy_voice.wav"


signal, sr = librosa.load(audio_path, sr=sr, mono = True)

print(signal)
```

```python
zcr = librosa.feature.zero_crossing_rate(signal)[0]
mean_zcr = np.mean(zcr)
std_zcr = np.std(zcr)
print(zcr)

print("*"*80)

print(f"MEAN ZCR : {mean_zcr: .2f}")
print(f"STD ZCR: {std_zcr: .2f}")
```

```python
import matplotlib.pyplot as plt

times  = librosa.times_like(zcr, sr=sr, hop_length=512)

plt.figure(figsize=(12, 4))
plt.plot(times, zcr, color = 'blue', linewidth = 1)
plt.xlabel('Time (Second)')
plt.ylabel('Zero Crossing Rate')
plt.title('ZCR Over Time')
plt.grid(True, alpha = 0.3)
plt.tight_layout()
plt.show()
```
