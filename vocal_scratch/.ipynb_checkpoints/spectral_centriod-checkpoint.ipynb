## Spectral Centriod

```python
import numpy as np
```

```python

def spectral_centriod(x, sr = 16000):

    X = np.fft.rfft(x)

    magnitude = np.abs(X)

    frequency = np.arange(len(magnitude))

    centroid = np.sum(magnitude * frequency) / np.sum(magnitude)

    # Convert to HZ

    bin_width = sr / len(x)

    centroid_hz = centroid * bin_width

    return centroid_hz
```

```python
x = np.array([1.0, 0.5, -0.5, -1.0, -0.5, 0.5, 1.0, 0.5])
print(spectral_centriod(x, sr=16000))
```

```python
import librosa

sr = 16000

audio_path = "test_healthy_voice.wav"

audio, sr = librosa.load(audio_path, sr=sr , mono=True )
print(audio)
```

```python
spectral_centriod_= librosa.feature.spectral_centroid(y = audio, sr=sr)[0]
spectral_centriod_
```

```python
## Mean SC across all frames
float(np.mean(spectral_centriod_))
```
