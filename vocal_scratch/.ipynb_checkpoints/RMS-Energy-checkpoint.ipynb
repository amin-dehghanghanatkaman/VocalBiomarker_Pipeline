```python
import numpy as np 
import math
```

```python
# Method One (Very Data-structure-y)
def rmse(x):

    N = len(x)
    r = 0
    
    for i in x:
        value =(1/N)*(i**2)
        r =r + value
    return np.sqrt(r)
```

```python
x = [0.5, -0.5, 0.5, -0.5]
rmse(x)
```

```python
# Method Two (using more Numpy)

def rmse2(x):

    N = len(x)

    squared = np.pow(x, 2)
    sum_squared = np.sum(squared)
    r = np.sqrt((1/N) * sum_squared)

    return r
```

```python
rmse2(x)
```

```python
import os
import numpy as np
import librosa
import librosa.display
import matplotlib.pyplot as plt
import pandas as pd
from scipy.signal import find_peaks
import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully")
```

```python
sr = 6000

audio_path = "test_healthy_voice.wav"

audio, sr = librosa.load(audio_path, sr=sr , mono=True )
```

```python
rms = librosa.feature.rms(y = audio)[0]
rms_mean = np.mean(rms)
rms_std = np.std(rms)

print(f"RMS Energy is equal to: {rms}.")

print("=" *50)

print(f"The mean RMS is {rms_mean}")

print("=" *50)
print(f"The variation of RMS is {rms_std}")
```

```python
times  = librosa.times_like(rms, sr=sr, hop_length=512)

plt.figure(figsize=(12, 4))
plt.plot(times, rms, color = 'blue', linewidth = 1)
plt.xlabel('Time (Second)')
plt.ylabel('RMS Energy')
plt.title('RMS Energy Over Time')
plt.grid(True, alpha = 0.3)
plt.tight_layout()
plt.show()
```
