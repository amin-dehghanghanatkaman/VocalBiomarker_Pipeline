## Spectral Rollof

```python
import numpy as np
```

```python
def spectralRolloff(x, sr=1600, roll_thresh=0.85):

    X = np.fft.rfft(x)
    mag = np.abs(X)
    print(f"Magnitude spectrum |X| is: {mag}")
    print("*"*70)

    tse = np.sum(mag)
    print(f"Total Sum Energy is: {tse}")
    print("*"*70)

    ttse = tse * roll_thresh
    print(f"Threshold is: {ttse}")
    print("*"*70)


    cumulative = np.cumsum(mag)

    print(f"Cumulative Sum is : {cumulative}")
    print("*"*70)

    rolloff_bin = np.max(np.where(cumulative <= ttse)[0])

    print(f"rolloff bin is : {rolloff_bin}")
    print("*"*70)
    
    bin_width = sr / len(x)

    print(f"Bin Width is : {bin_width}")
    print("*"*70)
    
    return rolloff_bin * bin_width
```

```python
x = np.array([1.0, 0.5, -0.5, -1.0, -0.5, 0.5, 1.0, 0.5])
print(spectralRolloff(x, sr=16000))
```

```python
import librosa

sr = 16000

audio_path = "test_healthy_voice.wav"

audio, sr = librosa.load(audio_path, sr=sr , mono=True )
print(audio)
```

```python
spectral_Rolloff = librosa.feature.spectral_rolloff(y=audio, sr=sr)[0]
print(f"Spectral Rolloff is: {spectral_Rolloff}")

print("#"*80)

print(f"Mean Spectral Rolloff is: {np.mean(spectral_Rolloff)}")

```
